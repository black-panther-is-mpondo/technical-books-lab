````python
# [SETUP] Core imports and environment information

import sys
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Python:", sys.version.split()[0])
print("Pandas:", pd.__version__)
print("NumPy:", np.__version__)
print("Executable:", sys.executable)
````

````python
# [PATHS] Resolve repository paths so notebooks work offline
# This works whether Jupyter starts from the repo root or from a subfolder.

def find_repo_root(start: Path | None = None) -> Path:
    start = (start or Path.cwd()).resolve()

    for candidate in (start, *start.parents):
        if (
            (candidate / "pyproject.toml").exists()
            and (candidate / "books" / "pandas").exists()
        ):
            return candidate

    raise FileNotFoundError(
        "Could not find the technical-books-lab repository root. "
        "Open the notebook from somewhere inside the repository."
    )


REPO_ROOT = find_repo_root()
PANDAS_DIR = REPO_ROOT / "books" / "pandas"

RAW_DATA_DIR = PANDAS_DIR / "data" / "raw"
PROCESSED_DATA_DIR = PANDAS_DIR / "data" / "processed"
OUTPUT_DIR = PANDAS_DIR / "outputs"

DATASETS = {
    "vehicles": RAW_DATA_DIR / "vehicles.csv.zip",
    "alta": RAW_DATA_DIR / "alta-noaa-1980-2019.csv",
    "siena_presidents": RAW_DATA_DIR / "siena2018-pres.csv",
    "jetbrains_2020": RAW_DATA_DIR / "2020-jetbrains-python-survey.csv",
    "dirty_devil": RAW_DATA_DIR / "dirtydevil.txt",
    "hanksville": RAW_DATA_DIR / "hanksville.csv",
}

print("Repository:", REPO_ROOT)
print("Raw data:", RAW_DATA_DIR)
print("Processed data:", PROCESSED_DATA_DIR)
print("Outputs:", OUTPUT_DIR)
````

````python
# [OFFLINE CHECK] Verify the book datasets are available locally

missing = [name for name, path in DATASETS.items() if not path.exists()]

if missing:
    raise FileNotFoundError(
        "Missing offline dataset(s): " + ", ".join(missing)
    )

for name, path in DATASETS.items():
    size_mb = path.stat().st_size / (1024 ** 2)
    print(f"[READY] {name:<18} {size_mb:>7.2f} MB  ->  {path.name}")
````
