```python
# [BOOK EXAMPLE] A python example representing a table of counts of songs artists composed.

series = {
    "index": [0, 1, 2, 3],
    "data": [145, 142, 38, 13],
    "name": "songs",
}


def get(series, idx):
    value_idx = series["index"].index(idx)
    return series["data"][value_idx]


get(series, 1)
```

```python
# [BOOK EXAMPLE] A python example representing a table of counts of songs artists composed.
# Use string values for the index.

songs = {
    "index": [
        "Paul",
        "John",
        "George",
        "Ringo",
    ],
    "data": [145, 142, 38, 13],
    "name": "counts",
}

get(songs, "John")
```

```python
# [BOOK EXAMPLE] A pandas Series example representing a table of counts of songs artists composed.

import pandas as pd

songs2 = pd.Series(
    [
        145,
        142,
        38,
        13,
    ],
    name="counts",
)

songs2
```

The output above, the leftmost column is the index and the values in the index are axis-labels and the right column are the values of the series.

```python
# [BOOK EXAMPLE] A pandas Series example representing a table of counts of songs artists composed.
# Inspecting the index

songs2.index
```

```python
# [BOOK EXAMPLE] A pandas Series example representing a table of counts of songs artists composed.
# String-based index

songs3 = pd.Series(
    [
        145,
        142,
        38,
        13,
    ],
    name="counts",
    index=[
        "Paul",
        "John",
        "George",
        "Ringo",
    ],
)

songs3
```

```python
# [BOOK EXAMPLE] A pandas Series example representing a table of counts of songs artists composed.
# Inspecting the index of the string-based index

songs3.index
```

```python
# [BOOK EXAMPLE] To demonstrate that the values of the series (data) does not have to be numeric or homogenous.


class Foo:
    pass


ringo = pd.Series(
    [
        "Richard",
        "Starkey",
        13,
        Foo(),
    ],
    name="ringo",
)

ringo
```

```python
# [BOOK EXAMPLE] A pandas Series example representing a series that has a NaN(Not A Number) in it.
import numpy as np

nan_series = pd.Series(
    [
        2,
        np.nan,
    ],
    index=["Ono", "Clapton"],
)

nan_series
```

```python
# [BOOK EXAMPLE] A pandas Series example representing a series that has a NaN(Not A Number) in it.
# Nan ignored by the count method.

nan_series.count()
```

```python
# [BOOK EXAMPLE] A pandas Series example representing a series that has a NaN(Not A Number) in it.
# NInspecting the number of entries (including missing values) using the size property

nan_series.size
```

```python
# [BOOK EXAMPLE] A pandas Series example representing a series that has a NaN(Not A Number) in it.
# Optional Integer support for NaN

nan_series2 = pd.Series(
    [
        2,
        None,
    ],
    index=[
        "Ono",
        "Clapton",
    ],
    dtype="Int64",
)

nan_series2
```

```python
# [BOOK EXAMPLE] A pandas Series example representing a series that has a NaN(Not A Number) in it.
# Nan ignored by the count method.
# Operations on these still ignores NaN or <NA>

nan_series.count()
```

Most functions already use the 'Int64' so it is recommended to ignore the 'Int64' and clean up missing data.

```python
# [BOOK EXAMPLE] Categorical data in python

s = pd.Series(["m", "l", "xs", "s", "xl"], dtype="category")

s
```

```python
# [BOOK EXAMPLE] Categorical data in python
# Verifying if the categories are not ordered by default.

s.cat.ordered
```

```python
# [BOOK EXAMPLE] Categorical data in python
# Convert a non-categorical series to an ordered category.

s2 = pd.Series(["m", "l", "xs", "s", "xl"])
size_type = pd.api.types.CategoricalDtype(
    categories=["s", "m", "l"],
    ordered=True,
)
s3 = s2.astype(size_type)

s3
```

```python
s3 > "s"
```

## Key Takeaways

- The Series object is a one-dimensional data structure.
- The Series can hold numerical data, time data, strings or arbitrary python objects.
- The Series output may look 2 dimentional at first but the farless column in an axis it doesn't represent data.
- If you're dealing with numeric data, using pandas rather than a Python list will benefit you.
- Pandas is faster, consumes less memory, and comes with built-in methods that are very useful to manipulate data.
- The index abstraction allows for accessing values by position or label.
- A series can also have empty values and has some similaries to NomPy arrays.
- It is the primary workhorse of pandas, mastering it will pay dividends.
