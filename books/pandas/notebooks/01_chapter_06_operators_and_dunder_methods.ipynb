```python
# [BOOK EXAMPLE] 5.1 Loading the data 

import pandas as pd

vehicle_filename = DATASETS["vehicles"]

df = pd.read_csv(vehicle_filename)
city_mpg = df.city08
highway_mpg = df.highway08
```

```python
# [BOOK EXAMPLE] 6.2 Dunder methods: Calculating the average of the two series

(city_mpg + highway_mpg) / 2
```

```python
# [BOOK EXAMPLE] 6.3 Index alignment: Example of two series that have repeated index entries as well as non-common entries.

s1 = pd.Series(
    [10, 20, 30,],
    index=[1, 2, 2,],
    )
s2 = pd.Series(
    [35, 44, 53,],
    index=[2, 2, 4,], name="s2"
)
```

```python
s1
```

```python
s2
```

```python
s1 + s2
```

```python
# [BOOK EXAMPLE] 6.4 Broadcasting: Example of two series that have repeated index entries as well as non-common entries.
## Duplicate Index Alignment

s1.add(s2, fill_value=0)
```

```python
# [BOOK EXAMPLE] 6.6 Operator Methods: How NaNs are handled in Operator methods
## Example 1

s1 + s2
```

```python
# [BOOK EXAMPLE] 6.6 Operator Methods: How NaNs are handled in Operator methods
## Example 2

s1.add(s2)
```

```python
# [BOOK EXAMPLE] 6.6 Operator Methods: How NaNs are handled in Operator methods
## Example 3

s1.add(s2, fill_value=0)
```

```python
# [BOOK EXAMPLE] 6.7 Chaining: Use methods and instead of the operator.
## Example 1: Using operations

((city_mpg + highway_mpg)) / 2
```

```python
# [BOOK EXAMPLE] 6.7 Chaining: Use methods and instead of the operator.
## Example 2: Using methods

(city_mpg
    .add(highway_mpg)
    ).div(2)
```

## Key Takeaways

- Pandas respond to most common math operations.
- You can use operator directly and will broadcast the operation to all the values.
- Alternatively, you can also call the corresponding method for the operator to make chaning easier or parameterize the behavior of the operation.
