## Key Takeaways

- Pandas have two main data structures.
- The Series data structures is a 1-Dimensional data structure that can be seen as a column or a row.
-  The DataFrame structure is a 2-Dimensional data structure that can be seen as a table/spreadsheet with both rows and column.

## Questions / Things to Revisit

- Q: If you had a spreadsheet with data, which pandas data structure would you use to hold data? Why?
- A: DataFrame, because a dataframe with the spreadsheet analogy represent a single sheet, with columns and rows.

- Q: If you had a database with data, which pandas data structure would you use to hold the data? Why?
- A: DataFrame, because a dataframe with the database analogy represent a table, with columns and rows.
